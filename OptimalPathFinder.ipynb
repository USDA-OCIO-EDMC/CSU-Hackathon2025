```python
import matplotlib.pyplot as plt
# from matplotlib.animation import FuncAnimation
from robot_class import Robot
from TertiaryCostMap import CostMap
from A import Algorithm
from Visualizer import visualize


def main():
    # Generate the cost map
    grid_tert = CostMap.getTCM()

    # Run the shortest-path algorithm
    optimizedPath = Algorithm.startA(grid_tert)

    # Plot the path (static visualization)
    visualize.plotVisual(optimizedPath)

    # Initialize the robot
    bot = Robot(1)
    bot.set_path(optimizedPath)

    # Set up the matplotlib figure
    fig, ax = plt.subplots(figsize=(10, 7.5))
    ax.set_xlim(0, 10500)  # Adjust based on your coordinate system
    ax.set_ylim(0, 13200)
    ax.set_title('Robot Path Visualization')
    ax.grid(True)

    # Plot the full path
    if len(optimizedPath) > 1:
        x_coords, y_coords = zip(*optimizedPath)
        ax.plot(x_coords, y_coords, 'g-', linewidth=2, label='Path')

    # Initialize robot position marker
    robot_marker, = ax.plot([], [], 'ro', markersize=8, label='Robot')

    # Animation update function
    def update(frame):
        if frame < len(optimizedPath):
            bot.move_to_next()
            x, y = optimizedPath[frame]
            robot_marker.set_data([x], [y])
            return robot_marker,
        return []

    # Create animation
    # ani = FuncAnimation(fig, update, frames=len(optimizedPath),
    # interval=50, blit=True, repeat=False)

    # Add legend and show plot
    ax.legend()
    plt.show()


if __name__ == "__main__":
    main()
```
