```python
import os
import numpy as np
from skimage import io
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

def load_images_from_folder(folder):
    images = []
    for filename in os.listdir(folder):
        if filename.endswith('.tif'):
            img = io.imread(os.path.join(folder, filename))
            images.append(img)
    return images

def compute_iou(y_true, y_pred):
    # Flatten the arrays to 1D
    y_true_flat = y_true.flatten()
    y_pred_flat = y_pred.flatten()

    # Compute intersection and union
    intersection = np.sum(np.logical_and(y_true_flat, y_pred_flat))
    union = np.sum(np.logical_or(y_true_flat, y_pred_flat))
    
    # Compute IoU
    iou = intersection / union if union != 0 else 0
    return iou

# Paths to the directories
y_true_path = 'data/train/masks'
y_pred_path = 'output/RGB'

# Load images
y_true_images = load_images_from_folder(y_true_path)
y_pred_images = load_images_from_folder(y_pred_path)

# Assuming both directories have the same number of images and corresponding names
# Compare the true and predicted images
all_iou = []
for true_img, pred_img in zip(y_true_images, y_pred_images):
    # Convert the images to binary (road vs no-road)
    true_binary = (true_img > 0).astype(int)  # 1 for road, 0 for non-road
    pred_binary = (pred_img > 0).astype(int)  # 1 for predicted road, 0 for predicted non-road
    
    # Compute IoU for the current pair
    iou = compute_iou(true_binary, pred_binary)
    all_iou.append(iou)

    # Compute confusion matrix for this pair
    cm = confusion_matrix(true_binary.flatten(), pred_binary.flatten())
    
    # Plot confusion matrix for this image pair
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['No Road', 'Road'], yticklabels=['No Road', 'Road'])
    plt.title(f'Confusion Matrix for {y_true_path}')
    plt.xlabel('Predicted Labels')
    plt.ylabel('True Labels')
    plt.show()

# Compute the average IoU over all images
average_iou = np.mean(all_iou)
print(f"Average IoU over all images: {average_iou:.4f}")

# Optionally, you can also compute the overall confusion matrix
y_true_all = np.concatenate([((img > 0).astype(int)).flatten() for img in y_true_images])
y_pred_all = np.concatenate([((img > 0).astype(int)).flatten() for img in y_pred_images])

overall_cm = confusion_matrix(y_true_all, y_pred_all)
plt.figure(figsize=(6, 5))
sns.heatmap(overall_cm, annot=True, fmt='d', cmap='Blues', xticklabels=['No Road', 'Road'], yticklabels=['No Road', 'Road'])
plt.title('Overall Confusion Matrix')
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')
plt.show()
```

```python
import os
import numpy as np
from skimage import io
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

def load_images_from_folder(folder, filenames):
    images = []
    for filename in filenames:
        img_path = os.path.join(folder, filename)
        if os.path.exists(img_path):
            img = io.imread(img_path)
            images.append(img)
        else:
            print(f"File not found: {img_path}")  # Print missing files
    return images

def compute_f1_score(y_true, y_pred):
    # Flatten the arrays to 1D
    y_true_flat = y_true.flatten()
    y_pred_flat = y_pred.flatten()

    # Compute confusion matrix with specified labels to ensure both classes are represented
    cm = confusion_matrix(y_true_flat, y_pred_flat, labels=[0, 1])

    # Extract TP, FP, FN, TN
    TP = cm[1, 1]  # True positives
    FP = cm[0, 1]  # False positives
    FN = cm[1, 0]  # False negatives
    TN = cm[0, 0]  # True negatives

    # Calculate Precision and Recall
    precision = TP / (TP + FP) if (TP + FP) != 0 else 0
    recall = TP / (TP + FN) if (TP + FN) != 0 else 0

    # Calculate F1 Score
    f1 = 2 * (precision * recall) / (precision + recall) if (precision + recall) != 0 else 0
    return f1

# Paths to the directories
y_true_path = 'data/test/masks'
y_pred_path = 'output/RGB'

# Get the list of filenames from the predicted directory (RGB images)
y_pred_filenames = sorted([f for f in os.listdir(y_pred_path) if f.endswith('.tif')])

# Create a new list of filenames for the mask images by altering the names
y_true_filenames = []
for filename in y_pred_filenames:
    # Alter the RGB filename to match the mask format
    new_filename = filename.replace("pred_mask", "SCC_mask")
    y_true_filenames.append(new_filename)

# Load the images based on the generated filenames
y_pred_images = load_images_from_folder(y_pred_path, y_pred_filenames)
y_true_images = load_images_from_folder(y_true_path, y_true_filenames)

# Initialize the overall confusion matrix to zero
overall_cm = np.zeros((2, 2))

# To store the flattened binary values for F1 score calculation
all_f1_scores = []

# Compare the true and predicted images
for true_img, pred_img in zip(y_true_images, y_pred_images):
    # Convert the images to binary (road vs no-road)
    true_binary = (true_img > 0).astype(int)  # 1 for road, 0 for non-road
    pred_binary = (pred_img > 0).astype(int)  # 1 for predicted road, 0 for predicted non-road
    
    # Compute F1 score for the current pair
    f1 = compute_f1_score(true_binary, pred_binary)
    all_f1_scores.append(f1)

    # Compute confusion matrix for this pair
    cm = confusion_matrix(true_binary.flatten(), pred_binary.flatten(), labels=[0, 1])
    
    # Add the current confusion matrix to the overall confusion matrix
    overall_cm += cm

# Compute the average F1 score over all images
average_f1 = np.mean(all_f1_scores)
print(f"Average F1 Score over all images: {average_f1:.4f}")

# Compute the average F1 minus 1 score over all images
average_f1_minus_1 = np.mean(np.array(all_f1_scores) - 1)
print(f"Average F1 - 1 Score over all images: {average_f1_minus_1:.4f}")

# Plot the overall confusion matrix
plt.figure(figsize=(6, 5))
sns.heatmap(overall_cm.astype(int), annot=True, fmt='d', cmap='Blues', xticklabels=['No Road', 'Road'], yticklabels=['No Road', 'Road'])
plt.title('Overall Confusion Matrix')
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')
plt.show()
```
