```python
import rasterio
import geopandas as gpd
import matplotlib.pyplot as plt

from PIL import Image
from rasterio.plot import show


shapefile = r"South_Clear_Creek/Roads_Boundary/South_Clear_Creek_Roads.shp"
df = gpd.read_file(shapefile)

demfile = r"South_Clear_Creek/Lidar_DEM_Hillshade/South_Clear_Creek_BareEarth_Hillshade_1m.tif"
src = rasterio.open(demfile)

fig, ax = plt.subplots(figsize=(10, 10))

df.plot(ax=ax, color="red", linewidth=2)

show(src, cmap="gray", ax=ax)

ax.set_title("DEM Overlay with Boundary")

output_png = "output_plot.png"
plt.savefig(output_png, format="png")
plt.show()

png_filepath = 'output_plot.png'
img = Image.open(png_filepath)
tiff_filepath = 'plot.tif'
img.save(tiff_filepath, format="TIFF")
```
